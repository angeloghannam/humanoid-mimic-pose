```python
from camera import Camera
import cv2

cam = Camera()

while True:

    frame = cam.read()

    cv2.imshow("camera", frame)

    if cv2.waitKey(1) == 27:
        break
```

```python
def normalize_to_pixel(keypoints, frame):

    h, w, _ = frame.shape

    keypoints[:,0] *= w
    keypoints[:,1] *= h

    return keypoints
```
