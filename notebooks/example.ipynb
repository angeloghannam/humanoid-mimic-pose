```python
from vision.camera import Camera
import cv2

cam = Camera()

while True:

    frame = cam.read()
    # If no camera is found, cam.read() will return None, 
    # so we check for that before trying to display the frame
    if frame is None:
        print("No camera found")
        break

    cv2.imshow("camera", frame)

    if cv2.waitKey(1) == 27:
        break
```

```python
def normalize_to_pixel(keypoints, frame):

    h, w, _ = frame.shape

    keypoints[:,0] *= w
    keypoints[:,1] *= h

    return keypoints
```
